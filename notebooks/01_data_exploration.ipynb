# 📊 01 — Data Exploration

Explore the pancreatic cancer CT scan dataset:
- Dataset statistics (image count per class)
- Sample image visualization
- Class distribution
- Image properties (resolution, intensity)

```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from PIL import Image

import config
from src.data.dataset import load_dataset_info, get_sample_images
from src.data.preprocessing import get_image_stats

%matplotlib inline
plt.style.use('dark_background')
print('✓ Libraries loaded')
```

## 1. Setup Directories
Ensure all project directories exist.

```python
config.setup_directories()
print(f'Data directory: {config.DATA_DIR}')
print(f'Train directory: {config.TRAIN_DIR}')
print(f'Val directory: {config.VAL_DIR}')
print(f'Test directory: {config.TEST_DIR}')
```

## 2. Dataset Statistics
Check how many images exist in each class and split.

```python
info = load_dataset_info()
print('Dataset Information:')
print('=' * 40)
for split, classes in info.items():
    if split == 'total':
        print(f'\nTotal images: {classes}')
    else:
        print(f'\n{split.upper()}:')
        for cls, count in classes.items():
            print(f'  {cls}: {count} images')
```

## 3. Class Distribution

```python
# Plot class distribution per split
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
colors = ['#00cec9', '#e17055']

for idx, split in enumerate(['train', 'val', 'test']):
    data = info.get(split, {})
    if data:
        classes = list(data.keys())
        counts = list(data.values())
        axes[idx].bar(classes, counts, color=colors[:len(classes)], edgecolor='white', linewidth=0.5)
        axes[idx].set_title(f'{split.upper()} Split', fontsize=13, fontweight='bold')
        axes[idx].set_ylabel('Image Count')
        for j, c in enumerate(counts):
            axes[idx].text(j, c + max(counts)*0.02, str(c), ha='center', fontweight='bold')
    else:
        axes[idx].text(0.5, 0.5, 'No data', ha='center', va='center', transform=axes[idx].transAxes)
        axes[idx].set_title(f'{split.upper()} Split')

fig.suptitle('Class Distribution Across Splits', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 4. Sample Images
Visualize sample images from each class.

```python
samples = get_sample_images(n_samples=5, split='train')

if samples:
    n_classes = len(samples)
    n_samples = max(len(v) for v in samples.values())
    
    if n_samples > 0:
        fig, axes = plt.subplots(n_classes, n_samples, figsize=(3*n_samples, 3*n_classes))
        if n_classes == 1:
            axes = [axes]
        
        for row, (cls, imgs) in enumerate(samples.items()):
            for col in range(n_samples):
                ax = axes[row][col] if n_classes > 1 else axes[col]
                if col < len(imgs):
                    ax.imshow(imgs[col])
                    if col == 0:
                        ax.set_ylabel(cls, fontsize=12, fontweight='bold')
                ax.axis('off')
        
        fig.suptitle('Sample Images per Class', fontsize=14, fontweight='bold', y=1.01)
        plt.tight_layout()
        plt.show()
    else:
        print('No images found. Please add images to data/processed/train/Normal/ and data/processed/train/Cancerous/')
else:
    print('No samples found. Please prepare your dataset first.')
```

## 5. Image Properties
Analyze image resolutions and formats.

```python
stats = get_image_stats(str(config.TRAIN_DIR))
print(f'Total images: {stats["count"]}')
print(f'Mean size: {stats["mean_size"]}')
print(f'Formats: {stats["formats"]}')

if stats['sizes']:
    widths, heights = zip(*stats['sizes'])
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].hist(widths, bins=20, color='#6c5ce7', edgecolor='white', alpha=0.8)
    axes[0].set_title('Width Distribution', fontweight='bold')
    axes[0].set_xlabel('Width (px)')
    axes[1].hist(heights, bins=20, color='#00cec9', edgecolor='white', alpha=0.8)
    axes[1].set_title('Height Distribution', fontweight='bold')
    axes[1].set_xlabel('Height (px)')
    plt.tight_layout()
    plt.show()
```
